```python
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
    
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
movies['cast'][0]
```

```python
def convert3(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):
        if(counter !=3):
            L.append(i['name'])
            counter+=1;
        else :
            break
    return L
    
```

```python
movies['cast']=movies['cast'].apply(convert3)
```

```python
def fetch_directors(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L
```

```python
movies['crew']=movies['crew'].apply(fetch_directors)
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

## We have to remove spaces between the words for example ther are Sam Worthington and Sam Mendes , suppose I want to watch Sam Worthington movie , but the model searches for Sam Mendes as , their is Sam in both the the Words , so we will remove the space and make a word a single Entity so that the model does not get confuse when something is same btw them   

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])

```

```python
movies.head()
```

## Tags Column ==>> Concatenation of all the columns => to get a paragraph for each movie

```python
movies.drop(columns=['tags'], inplace=True)
```

```python
movies['tags']=movies['overview'] + movies['genres']+ movies['keywords']+ movies['cast']+ movies['crew']
```

```python


movies.head()
```

```python
new_df=movies[['movie_id','title','tags']]
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df['tags'][0]
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
```

```python
new_df.head()
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)

    
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

## To get Similarity between two texts /paras , we will be converting it to vector first using traditional NLP techniques

### We have  for a movie ==>  movie id \ movie names \ movie tags
### we will be converting each movie tags to a vector ==> such that each movie becomes a vector


## Such that vectors closest to that particular movie , will be recommended ,
## Also we will be taking max features 5000 (5000 most common words), means 5000 vectors 

# TEXT ===> VECTORS

### We will be using Bag of Words 

### Combine all movies tags -> tags1 + tags2 + tags3 + ---- + tags49529
### From the -> large text ==> Extract 5000 most common words ==> w1 \ w2 \ w3 \ .....\ w5000
### Now in Every Movie Tag => Search how many common words are there from these 5000 most common words 
### Example in movie1  tag => w1 was there 5 times , w2 was there 3 times , w3 was present 0 times  like this 

### This will happen for all the movies , m1 , m2 ,--- , m49529
### So for each movie a 5000 length size vector will be created storing each word frequncy from those 5000 words , the words which are new to , not from the 5000 common words are given a value 0

## Demo -> m1 =[5,3,2,7,0,1,4,9,0,0,1.....](frequency -> (w1,w2,w3,---w5000))

## While doing this we will remove stop words

```python
from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
def recommend(movie):
    movie_index=new_df[new_df['title']==movie].index[0]
    distances=similarity[movie_index]
    movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
    
    
```

```python
recommend('Batman Begins')
```
